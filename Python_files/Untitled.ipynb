```python
import pandas as pd
from faker import Faker
import random
from datetime import datetime
```

```python
fake = Faker()
```

```python
data = []

for _ in range(20):
    
    dob = fake.date_of_birth(minimum_age=18, maximum_age=80)
    
    start_date = datetime(dob.year + 18, 1, 1)

    created_at = fake.date_time_between(
        start_date=start_date
    )
    
    record = {
        'customer_id': fake.unique.random_number(digits=6),
        'name': fake.name(),
        'email': fake.unique.email(),
        'phone': fake.phone_number(),
        'address': fake.address().replace('\n', ', '),
        'dob': dob,
        'created_at': created_at
    }
    
    data.append(record)

df = pd.DataFrame(data)

print(df.head(15))
```

```python
branch_names = [
    "Central Branch", "Lake View Branch", "Park Street Branch",
    "City Center Branch", "Hill Road Branch", "Main Market Branch",
    "Riverside Branch", "Green Park Branch", "Sunrise Branch",
    "Metro Branch", "Airport Branch", "Industrial Area Branch",
    "College Road Branch", "New Town Branch", "South City Branch"
]

cities = [
    "New York", "London", "Tokyo", "Paris", "Sydney",
    "Dubai", "Singapore", "Toronto", "Berlin", "Rome",
    "Bangkok", "Seoul", "Amsterdam", "Barcelona", "Istanbul"
]

data = []

for i in range(1, 16):

    record = {
        "branch_id": i,
        "branch_name": branch_names[i-1],
        "city": cities[i-1],
        "ifsc_code": "BANK0" + str(random.randint(100000, 999999))
    }

    data.append(record)

df = pd.DataFrame(data)

print(df)
```

```python
roles = [
    "Manager", "Cashier", "Accountant", "Clerk",
    "Loan Officer", "HR Executive", "Security Officer",
    "Branch Head", "Customer Support", "Auditor"
]

data = []

for i in range(1, 16):

    record = {
        "employee_id": i,
        "branch_id": random.randint(1, 15),
        "name": fake.name(),
        "role": random.choice(roles),
        "salary": random.randint(30000, 120000)
    }

    data.append(record)

df = pd.DataFrame(data)

print(df)
```

```python
customer_ids = [
    86424, 264994, 869320, 593386, 918656,
    827624, 770953, 596838, 176740, 520143,
    313991, 504080, 658893, 349285, 789565
]

loan_types = [
    "Home Loan",
    "Car Loan",
    "Education Loan",
    "Personal Loan",
    "Business Loan",
    "Gold Loan"
]

statuses = [
    "Active",
    "Closed",
    "Pending",
    "Defaulted"
]

data = []

for i in range(1, 16):

    record = {
        "loan_id": 2000 + i,
        "customer_id": customer_ids[i - 1],
        "loan_type": random.choice(loan_types),
        "principal_amount": round(random.uniform(50000, 5000000), 2),
        "interest_rate": round(random.uniform(6.5, 15.0), 2),
        "tenure_months": random.choice([12, 24, 36, 60, 84, 120, 180, 240]),
        "status": random.choice(statuses)
    }

    data.append(record)

df = pd.DataFrame(data)

print(df)
```

```python
account_ids = [
    1001, 1002, 1003, 1004, 1005,
    1006, 1007, 1008, 1009, 1010,
    1011, 1012, 1013, 1014, 1015
]

descriptions = [
    "ATM Withdrawal",
    "Online Transfer",
    "Salary Credit",
    "Bill Payment",
    "Loan EMI",
    "Cash Deposit",
    "UPI Payment",
    "Cheque Deposit",
    "Shopping Purchase",
    "Interest Credit"
]

data = []

for i in range(1, 16):

    amount = round(random.uniform(500, 50000), 2)

    balance_after = round(random.uniform(1000, 1000000), 2)

    txn_time = fake.date_time_between(
        start_date='-2y',
        end_date='now'
    )

    record = {
        "txn_id": 3000 + i,
        "account_id": account_ids[i - 1],
        "amount": amount,
        "balance_after": balance_after,
        "txn_time": txn_time,
        "description": random.choice(descriptions)
    }

    data.append(record)

df = pd.DataFrame(data)

print(df)
```

```python
account_ids = [
    1001, 1002, 1003, 1004, 1005,
    1006, 1007, 1008, 1009, 1010,
    1011, 1012, 1013, 1014, 1015
]

card_types = [
    "Debit Card",
    "Credit Card",
    "Platinum Credit Card",
    "Business Credit Card",
    "Student Debit Card"
]

data = []

for i in range(1, 16):

    expiry_date = fake.credit_card_expire(start="now", end="+5y")

    record = {
        "card_id": 4000 + i,
        "account_id": account_ids[i - 1],
        "card_type": random.choice(card_types),
        "card_number": fake.unique.credit_card_number(),
        "expiry": expiry_date,
        "credit_limit": round(random.uniform(10000, 500000), 2)
    }

    data.append(record)

df = pd.DataFrame(data)

print(df)
```
